```python
# operadores lógicos
numero1 = 1
numero2 = 2

# exemplo de if com operador lógico AND
if numero1 == 1 and numero2 == 2:
  print("Os valores são iguais!")
```

```python
# exemplo de if com operador lógico OR
if numero1 > 0 or numero2 < 0:
  print("O número 1 é maior que 0, " +
          "mas o número 2 não é menor que 0")
```

```python
# exemplo de if com operdor lógico de NEGAÇÃO
if not numero2 < 0:
  print("O número 2 é maior que 0")
```

```python
# exemplo de if com else
valor1 = 3

if valor1 >= 3:
  print("O valor 1 é maior ou igual à 3")
else:
  print("O valor 1 não é maior ou igual à 3")
```

```python
# exemplo de if com elif e else
idade = 61

if idade < 18:
  print("Jovem")
elif idade >= 18 and idade <= 60:
  print("Adulto")
else:
  print("Idoso")
```

```python
# exemplo de if abreviado
dia = 10

if dia == 10: print("Hoje é dia 10") # apenas if
print("Hoje é dia 10") if dia == 10 else print("Hoje não é dia 10")
```

```python
# exemplo match/case
opcao = int(input("Informe a opção: "))

match opcao:
  case 1:
    print("Falar com atendente")
  case 2:
    print("Dúvidas Frequentes")
  case _:
    print("Opção inválida")
```

```python
# estruturas de repetição
# exemplo simples de while
contador = 0

while(contador < 10):
  print(contador)

  # vamos forçar a parada
  if contador == 4:
    break

  contador += 1
```

```python
# exemplo de repetição com continue
# pular uma execução
contador = 0

while (contador < 10):
  contador += 1

  # pular a interação
  if contador == 2:
    continue

  print(contador)
```

```python
# exemplo de while com else
contador = 0

while (contador < 10):
  print(contador)
  contador += 1
else:
  print("O contador é maior do que 10")
```

```python
# exemplo de lista
lista = ["Ana", "Paula", "Mariana"]
print(type(lista)) # tipo list

for nome in lista:
  print(nome)
```

```python
# exemplo de for em uma string
frase = "o palmeiras não tem mundial!"

for c in frase:
  print(c)
```

```python
# exemplo de uma lista
# com for e break
lista = [1,2,3,4,5]

for num in lista:
  print(num)

  if num == 3:
    break
```

```python
# exemplo de adicionar um item na lista
lista = ["Londrina", "Maringá", "Toledo"]
lista.append("Cascavel") # adiciona no final
print(lista)
lista.insert(0, "Turvo")
print(lista)

# exemplo de busca em uma lista
if "Toledo" in lista:
  print("existe a cidade mencionada acima")
```

```python
# exemplo de contagem de qtd de items na lista
print(len(lista))

# visualizar um item escolhido
print(lista[0])
print(lista[-1]) # ultimo item
print(lista[:2]) # visualizando 2 valores começando do inicio
print(lista[3:]) # visualizando 2 valores começando do final
```

```python
# exemplo de atualizar valores
lista = [1,2,3,4,5]
print(lista)
lista[0] = 10
print(lista)

# exemplo de remover items
lista.remove(10)
print(lista)
lista.pop(2)
print(lista)

# exemplo de limpar a lista
lista.clear()
# lista = []
print(lista)
```

```python
# exemplo de uma ordenação de lista
lista = ["Mateus", "Zeus", "Adriano"]
lista.sort() # crescente/alfabetica
print(lista)
lista.sort(reverse = True) # decrescente
print(lista)
```

```python
# exemplo junção de listas
lista1 = [2,2,2]
lista2 = [4,5,6]
lista3 = [7]

lista_final = lista1 + lista2 + lista3
print(lista_final)
lista1.extend(lista2)
lista1.extend(lista3)
print(lista1)

print(lista1.count(2)) # contagem de items por valor
```
